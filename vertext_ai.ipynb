```python
import io
import mimetypes
from google.cloud import storage
from google.cloud import discoveryengine_v1
from PyPDF2 import PdfReader

#RAG to bucket
def fetch_gcs_file(uri: str) -> str:
    """Ambil isi file dari GCS (support banyak file, pdf & text)."""
    if not uri.startswith("gs://"):
        return ""

    parts = uri.replace("gs://", "").split("/", 1)
    bucket_name = parts[0]
    prefix = parts[1] if len(parts) > 1 else ""

    client = storage.Client()
    bucket = client.bucket(bucket_name)

    # Ambil semua file dengan prefix (bisa 1 file, bisa folder)
    blobs = bucket.list_blobs(prefix=prefix)
    print(blobs)
    texts = []
    for blob in blobs:
        mime_type, _ = mimetypes.guess_type(blob.name)

        if mime_type == "application/pdf":
            # Ekstrak PDF
            pdf_bytes = blob.download_as_bytes()
            reader = PdfReader(io.BytesIO(pdf_bytes))
            text = ""
            for page in reader.pages:
                if page.extract_text():
                    text += page.extract_text() + "\n"
            if text.strip():
                texts.append(f"=== {blob.name} ===\n{text.strip()}")

        else:
            # Asumsikan text file
            data = blob.download_as_bytes()
            try:
                decoded = data.decode("utf-8", errors="ignore")
                texts.append(f"=== {blob.name} ===\n{decoded.strip()}")
            except Exception as e:
                texts.append(f"=== {blob.name} ===\n[ERROR membaca file: {e}]")

    return "\n\n".join(texts)


@tool
def tool_knowledge_base(query: str) -> str:
    """
    Gunakan untuk menjawab pertanyaan dari datastore (Standard Edition).
    Ambil konten dari Discovery Engine + GCS bucket.
    """
    client = discoveryengine_v1.SearchServiceClient()
    serving_config = (
        f"projects/{PROJECT_ID}/locations/global/dataStores/{DATA_STORE_ID}/servingConfigs/default_config"
    )
    request = discoveryengine_v1.SearchRequest(
        serving_config=serving_config, query=query, page_size=5
    )
    response = client.search(request=request)
    results = []
    for result in response:
        doc = result.document
        link = doc.derived_struct_data.get("link", "")

        if link.startswith("gs://"):
            text = fetch_gcs_file(link)
        else:
            # fallback: pakai snippet dari discovery engine
            text = result.snippet.snippet_text if result.snippet else ""

        if text:
            results.append(text)
    hasil = "\n\n".join(results) if results else "Tidak ada hasil relevan."
    return hasil
```

```python
# call api third party
import requests
from langchain_core.tools import tool

@tool
def get_dummy_api_response(query) -> dict:
    """
    Digunakan untuk informasi data lokasi dari api berdasarkan query user
    """
    url = "https://third-party-api-service-YOUR_PROJECT_NUMBER.YOUR_REGION.run.app/locations"
    params={
            "q": query,
            "location": "Austin, Texas, United States",
            "hl": "en",
            "gl": "us",
            "google_domain": "google.com",
            "api_key": "YOUR_UPSTREAM_API_KEY"
            }
    print("query = ",query)
    try:
        response = requests.get(url, params=params, timeout=10)
        return response.json()
    except requests.RequestException as e:
        print(e)
        return {"error": str(e)}
```

```python
store = {}

def get_session_history(session_id:str)->BaseChatMessageHistory:
  if session_id not in store:
    store[session_id]=ChatMessageHistory()
  return store[session_id]
```

```python
# setting up agent
model = LLM_MODEL
agent = reasoning_engines.LangchainAgent(
    model=model,
    tools=tools,
    prompt=prompt,
    chat_history=get_session_history,
    agent_executor_kwargs={
        "return_intermediate_steps":True
    },
    model_kwargs=model_kwargs,
    enable_tracing=True
)
```
